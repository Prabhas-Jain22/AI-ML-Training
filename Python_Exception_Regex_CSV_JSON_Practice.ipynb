## 1. Exception Handling

### Theory
An **exception** is an error raised while a program is running. Exception handling lets a program respond to expected failures instead of stopping unexpectedly.

| Keyword | Purpose |
|---|---|
| `try` | Contains code that may raise an exception |
| `except` | Handles a matching exception |
| `else` | Runs only when the `try` block succeeds |
| `finally` | Runs whether an exception occurred or not |
| `raise` | Raises an exception intentionally |

Catch specific exceptions where possible. For example, use `ValueError` for invalid conversions and `ZeroDivisionError` for division by zero. Avoid a bare `except:` because it can hide programming errors.

### Sample example
The user may enter text instead of a number, so converting input should be protected.

```python
# Sample: safe numeric conversion and division
def divide_values(numerator_text, denominator_text):
    try:
        numerator = float(numerator_text)
        denominator = float(denominator_text)
        result = numerator / denominator
    except ValueError:
        print("Please provide valid numeric values.")
    except ZeroDivisionError:
        print("The denominator cannot be zero.")
    else:
        print(f"Result: {result}")
    finally:
        print("Operation finished.")

divide_values("12", "3")
divide_values("twelve", "3")
divide_values("12", "0")
```

## 2. Regular Expressions (Regex)

### Theory
A **regular expression** is a pattern used to search, validate, split, or replace text. Python provides regex support through the built-in `re` module.

| Pattern | Meaning |
|---|---|
| `\d` | A digit |
| `\w` | A word character (letters, digits, underscore) |
| `\s` | Whitespace |
| `.` | Any character except a newline by default |
| `+` | One or more repetitions |
| `*` | Zero or more repetitions |
| `{n}` | Exactly `n` repetitions |
| `^` / `$` | Start / end of string |
| `[A-Z]` | Any uppercase English letter |
| `(...)` | Capturing group |

Common functions:
- `re.search()` finds the first match anywhere in a string.
- `re.findall()` returns all non-overlapping matches.
- `re.fullmatch()` checks whether the entire string matches.
- `re.sub()` replaces matching text.

Use raw strings such as `r"\d+"` for regex patterns so backslashes are handled clearly. Regex can check a basic email-like pattern, but it cannot guarantee that an email address exists or is deliverable.

### Sample example

```python
import re

text = "Orders: 1042, 1043, and 1088. Contact: support@example.com"
order_ids = re.findall(r"\b\d{4}\b", text)
emails = re.findall(r"[\w.-]+@[\w.-]+\.\w+", text)

print("Order IDs:", order_ids)
print("Email-like matches:", emails)

# Replace multiple spaces with a single space
messy_text = "Python     makes   data work easier."
clean_text = re.sub(r"\s+", " ", messy_text).strip()
print(clean_text)
```

## 3. File Handling: CSV

### Theory
CSV (**Comma-Separated Values**) is a common tabular format. Each row represents a record, and columns represent fields.

Python's built-in `csv` module supports CSV files:
- `csv.DictReader` reads rows as dictionaries keyed by column names.
- `csv.DictWriter` writes dictionaries to a CSV file.
- `newline=""` is recommended when opening CSV files for reading or writing.
- `encoding="utf-8"` helps handle a broad range of text.

Use a context manager (`with open(...)`) so the file is closed automatically. CSV values are generally read as strings; convert values such as quantities or prices when calculations are needed.

### Sample example
This cell creates a small CSV file, reads it, and calculates the total sales value. It creates the sample file in the current Colab session.

```python
import csv

csv_file = "sales.csv"
sales_rows = [
    {"order_id": "O101", "product": "Notebook", "quantity": "3", "unit_price": "50"},
    {"order_id": "O102", "product": "Pen", "quantity": "10", "unit_price": "12"},
    {"order_id": "O103", "product": "Bag", "quantity": "2", "unit_price": "650"},
]

with open(csv_file, "w", newline="", encoding="utf-8") as file:
    fieldnames = ["order_id", "product", "quantity", "unit_price"]
    writer = csv.DictWriter(file, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(sales_rows)

with open(csv_file, "r", newline="", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    total_sales = 0
    for row in reader:
        line_total = int(row["quantity"]) * float(row["unit_price"])
        total_sales += line_total
        print(row["order_id"], row["product"], "line total =", line_total)

print("Total sales value =", total_sales)
```

## 4. File Handling: JSON

### Theory
JSON (**JavaScript Object Notation**) is a text format commonly used for APIs and nested data. It supports objects, arrays, strings, numbers, booleans, and `null`.

Python's built-in `json` module provides:
- `json.load(file)` to read JSON from an open file.
- `json.dump(data, file, indent=2)` to write Python data to a file in readable form.
- `json.loads(text)` to parse a JSON string.
- `json.dumps(data)` to convert Python data into a JSON string.

A JSON object usually becomes a Python dictionary, while a JSON array becomes a Python list. Invalid JSON raises `json.JSONDecodeError`.

### Sample example

```python
import json

json_file = "inventory.json"
inventory = {
    "warehouse": "North",
    "last_updated": "2026-10-11",
    "items": [
        {"sku": "SKU101", "name": "Keyboard", "stock": 25},
        {"sku": "SKU102", "name": "Mouse", "stock": 8},
        {"sku": "SKU103", "name": "Monitor", "stock": 4}
    ]
}

# Write Python data as JSON
with open(json_file, "w", encoding="utf-8") as file:
    json.dump(inventory, file, indent=2)

# Read JSON back into Python objects
with open(json_file, "r", encoding="utf-8") as file:
    loaded_inventory = json.load(file)

for item in loaded_inventory["items"]:
    print(f"{item['name']}: {item['stock']} units")

low_stock = [
    item["name"] for item in loaded_inventory["items"]
    if item["stock"] < 10
]
print("Items with low stock:", low_stock)
```

## 5. Combining File Handling with Exception Handling

Files may be missing, unreadable, or malformed. Handle expected failures explicitly.

- `FileNotFoundError`: the file path does not exist.
- `PermissionError`: the program lacks permission to access the file.
- `json.JSONDecodeError`: the file contents are not valid JSON.
- `csv.Error`: a CSV parsing or writing issue occurred.

### Sample example

```python
import json

def load_json_safely(path):
    try:
        with open(path, "r", encoding="utf-8") as file:
            data = json.load(file)
    except FileNotFoundError:
        print(f"File not found: {path}")
        return None
    except json.JSONDecodeError as error:
        print(f"Invalid JSON: {error}")
        return None
    else:
        print("JSON loaded successfully.")
        return data

result = load_json_safely("inventory.json")
print("Warehouse:", result["warehouse"] if result else "Unavailable")

# Demonstrate a missing-file case
load_json_safely("missing_file.json")
```
